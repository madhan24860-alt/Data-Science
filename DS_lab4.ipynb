```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
df=pd.read_csv("iris.csv")
df
```

```python
stats.probplot(
    df["sepal.length"],
    dist="norm",
    plot=plt
)
plt.title("normal distribution QQ plot - iris sepal lenght")
plt.show()
```
