```python
import numpy as np
import pandas as pd
c=pd.read_csv("HollywoodMovies.csv")
c
```

```python
data=c["Profitability"]
data
```

```python
from scipy.stats import skew
ans=skew(data)
if ans<0:
    print("Negative skew")
elif ans>0:
    print("Positive skew")
else:
    print("Normal skew")
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.hist(data, color="red", bins=10)
plt.title("profit distribution")
plt.xlabel("Profitability")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```
